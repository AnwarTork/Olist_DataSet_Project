```python
from pathlib import Path
import sys

import logging

logger = logging.getLogger(__name__)
PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
import sys 
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    stream=sys.stdout
)
```

```python
import yaml

from src.labeling import (
    create_label_pipeline,
    get_label_distribution,
    calculate_imbalance_ratio,
    validate_labeled_table,
)
```

```python
with open(
    PROJECT_ROOT / "config" / "config.yaml",
    "r",
    encoding="utf-8",
) as file:
    config = yaml.safe_load(file)
```

```python
input_path = (
    PROJECT_ROOT
    / config["paths"]["ml_table"]
)

output_path = (
    PROJECT_ROOT
    / config["paths"]["labeled_table"]
)

logger.info("Input: %s", input_path)
logger.info("Output:%s", output_path)
```

```python
result = create_label_pipeline(
    input_path=input_path
)

labeled_table = result["data"]

logger.info("Original rows: %s", result["original_rows"])
logger.info("Labeled rows: %s", result["labeled_rows"])
logger.info("Removed rows: %s", result["removed_rows"])
logger.info(
    "All labels correct: %s",
    result["labels_correct"]
)
```

```python
distribution = get_label_distribution(
    labeled_table
)

distribution
```

```python
imbalance = calculate_imbalance_ratio(
    labeled_table
)

logger.info(
    "Majority class:%s",
    imbalance["majority_count"]
)

logger.info(
    "Minority class:%s",
    imbalance["minority_count"]
)

logger.info(
    "Imbalance ratio:%s",
    round(
        imbalance["imbalance_ratio"],
        2
    )
)
```

```python
validation = validate_labeled_table(
    labeled_table
)

validation
```

```python
logger.info("Rows:%s", validation["rows"])
logger.info(
    "Unique orders:%s",
    validation["unique_orders"]
)
logger.info(
    "Duplicate orders:%s",
    validation["duplicate_orders"]
)
logger.info(
    "Missing labels:%s",
    validation["missing_labels"]
)
```
