```python
items_agg = (
    order_items
    .groupby("order_id")
    .agg(
        total_items=("order_item_id", "count"),
        total_price=("price", "sum"),
        total_freight_value=("freight_value", "sum"),
        avg_item_price=("price", "mean")
    )
    .reset_index()
)
```

```python
items_agg.head()
```

```python
payments_agg = (
    order_payments
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        max_installments=("payment_installments", "max"),
        payment_count=("payment_sequential", "count")
    )
    .reset_index()
)
```

```python
payments_agg.head()
```

```python
payment_type = (
    order_payments
    .groupby("order_id")["payment_type"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
    .reset_index()
    .rename(columns={"payment_type": "main_payment_type"})
)
payment_type.head()
```

```python
payments_agg = payments_agg.merge(
    payment_type,
    on="order_id",
    how="left"
)
```

```python
reviews_agg = (
    order_reviews
    .groupby("order_id")
    .agg(
        review_score=("review_score", "mean"),
        review_count=("review_id", "count")
    )
    .reset_index()
)
reviews_agg.head()
```

```python
products["product_volume_cm3"] = (
    products["product_length_cm"]
    * products["product_height_cm"]
    * products["product_width_cm"]
)
```

```python
order_product_data = order_items[
    [
        "order_id",
        "product_id"
    ]
].merge(
    products[
        [
            "product_id",
            "product_weight_g",
            "product_volume_cm3"
        ]
    ],
    on="product_id",
    how="left"
)
```

```python
product_features = (
    order_product_data
    .groupby("order_id")
    .agg(
        total_weight=("product_weight_g", "sum"),
        total_volume=("product_volume_cm3", "sum"),
        avg_volume=("product_volume_cm3", "mean"),
        max_volume=("product_volume_cm3", "max")
    )
    .reset_index()
)
```

```python
ml_table = ml_table.merge(
    product_features,
    on="order_id",
    how="left"
)
```

```python
seller_order_data = (
    order_items[
        [
            "order_id",
            "seller_id"
        ]
    ]
    .merge(
        sellers[
            [
                "seller_id",
                "seller_zip_code_prefix",
                "seller_state"
            ]
        ],
        on="seller_id",
        how="left"
    )
)
```

```python
geo_by_zip = (
    geolocation
    .groupby("geolocation_zip_code_prefix")
    .agg(
        latitude=("geolocation_lat", "mean"),
        longitude=("geolocation_lng", "mean")
    )
    .reset_index()
)
```

```python
customer_geo = ml_table[
    [
        "order_id",
        "customer_zip_code_prefix"
    ]
].merge(
    geo_by_zip,
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left"
)
```

```python
customer_geo = customer_geo.rename(
    columns={
        "latitude": "customer_lat",
        "longitude": "customer_lng"
    }
)
```

```python
seller_geo = seller_order_data.merge(
    geo_by_zip,
    left_on="seller_zip_code_prefix",
    right_on="geolocation_zip_code_prefix",
    how="left"
)
```

```python
seller_geo = seller_geo.rename(
    columns={
        "latitude": "seller_lat",
        "longitude": "seller_lng"
    }
)
```

```python
def haversine_distance(
    lat1,
    lon1,
    lat2,
    lon2
):
    R = 6371

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)

    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        +
        np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    c = 2 * np.arcsin(
        np.sqrt(a)
    )

    return R * c
```

```python
geo_orders = customer_geo.merge(
    seller_geo[
        [
            "order_id",
            "seller_id",
            "seller_state",
            "seller_lat",
            "seller_lng"
        ]
    ],
    on="order_id",
    how="left"
)
```

```python
geo_orders["distance_km"] = haversine_distance(
    geo_orders["customer_lat"],
    geo_orders["customer_lng"],
    geo_orders["seller_lat"],
    geo_orders["seller_lng"]
)
```

```python
geo_orders = geo_orders.merge(
    ml_table[
        [
            "order_id",
            "customer_state"
        ]
    ],
    on="order_id",
    how="left"
)
```

```python
geo_orders["is_cross_state"] = (
    geo_orders["customer_state"]
    != geo_orders["seller_state"]
).astype(int)
```

```python
geo_features = (
    geo_orders
    .groupby("order_id")
    .agg(
        distance_km=("distance_km", "mean"),
        is_cross_state=("is_cross_state", "max")
    )
    .reset_index()
)
```

```python
ml_table = ml_table.merge(
    geo_features,
    on="order_id",
    how="left"
)
```
