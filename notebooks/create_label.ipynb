```python
import pandas as pd
import numpy as np
```

```python
ml_table = pd.read_csv("C:\\Users\\Anwar Altorkmani\\Desktop\\Task2\\olist_ml_table_new_new_new.csv")

print("Shape:", ml_table.shape)
ml_table.head()
```

```python
date_columns = [
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

print(ml_table[date_columns].head())
```

```python
print(ml_table[date_columns].isna().sum())
```

```python
ml_table["order_delivered_customer_date"] = pd.to_datetime(
    ml_table["order_delivered_customer_date"],
    errors="coerce"
)

ml_table["order_estimated_delivery_date"] = pd.to_datetime(
    ml_table["order_estimated_delivery_date"],
    errors="coerce"
)
```

```python
print(ml_table[
    [
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].dtypes)
```

```python
labeled_table = ml_table[
    ml_table["order_delivered_customer_date"].notna() &
    ml_table["order_estimated_delivery_date"].notna()
].copy()
```

```python
print("Original rows:", len(ml_table))
print("Labeled rows:", len(labeled_table))
print("Removed rows:", len(ml_table) - len(labeled_table))
```

```python
labeled_table["is_late"] = (
    labeled_table["order_delivered_customer_date"]
    >
    labeled_table["order_estimated_delivery_date"]
).astype(int)
```

```python
labeled_table["delivery_delay_days"] = (
    labeled_table["order_delivered_customer_date"]
    -
    labeled_table["order_estimated_delivery_date"]
).dt.total_seconds() / 86400
```

```python
labeled_table[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_delay_days",
        "is_late"
    ]
].head(10)
```

```python
labeled_table[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_delay_days",
        "is_late"
    ]
].sample(10, random_state=42)
```

```python
labeled_table[
    labeled_table["is_late"] == 1
][
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_delay_days",
        "is_late"
    ]
].head(5)
```

```python
labeled_table[
    labeled_table["is_late"] == 0
][
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delivery_delay_days",
        "is_late"
    ]
].head(5)
```

```python
check = (
    (labeled_table["delivery_delay_days"] > 0).astype(int)
    == labeled_table["is_late"]
)

print("All labels correct:", check.all())
```

```python
class_counts = labeled_table["is_late"].value_counts()

print(class_counts)
```

```python
class_percentages = (
    labeled_table["is_late"]
    .value_counts(normalize=True)
    * 100
)

print(class_percentages)
```

```python
distribution = pd.DataFrame({
    "count": class_counts,
    "percentage": class_percentages
})

distribution.index = ["On Time (0)" if i == 0 else "Late (1)" 
                      for i in distribution.index]

distribution
```

```python
majority_count = labeled_table["is_late"].value_counts().max()
minority_count = labeled_table["is_late"].value_counts().min()

imbalance_ratio = majority_count / minority_count

print("Majority class:", majority_count)
print("Minority class:", minority_count)
print("Imbalance ratio:", round(imbalance_ratio, 2))
```

```python
print("Rows:", len(labeled_table))
print("Unique orders:", labeled_table["order_id"].nunique())
print(
    "Duplicate orders:",
    labeled_table["order_id"].duplicated().sum()
)
```

```python
labeled_table = labeled_table.drop(
    columns=["delivery_delay_days"]
)
```

```python
print(labeled_table.columns.tolist())
```

```python
print("Final shape:", labeled_table.shape)

print("\nLabel distribution:")
print(labeled_table["is_late"].value_counts())

print("\nMissing labels:")
print(labeled_table["is_late"].isna().sum())

print("\nDuplicate orders:")
print(labeled_table["order_id"].duplicated().sum())
```

```python
labeled_table.to_csv(
    "olist_labeled_ml_table_new.csv",
    index=False
)

print("Saved successfully!")
```
